```python
#Ejercicio 1
def contar_caracteres(palabra):
  numero_caracteres=0
  for e in palabra:
    numero_caracteres=numero_caracteres+1
  return numero_caracteres
```

```python
#Ejercicio 2
def calcular_promedio (lista):
  total=0
  if len(lista)==0:
    return 0
  for e in lista:
    total=total+e
  return total/len(lista)
```

```python
#Ejercicio 3
def encontrar_duplicado(lista):
  for e in lista:
    veces=0
    for i in lista:
      if e==i:
        veces=veces+1
    if veces==2:
      return e
    
```

```python
#Ejercicio 4
def enmascarado_datos(variable):
  texto=str(variable)
  if len(texto) <= 4:
    return texto
  final=""
  for e in range(len(texto)-4):
    final=final+"#"
  final=final+texto[-4:]
  return final
```

```python
#Ejercicio 5
def es_anagrama(palabra1,palabra2):
  if len(palabra1)!= len(palabra2):
    return "No es anagrama"
  elif sorted(palabra1.lower())==sorted(palabra2.lower()):
    return "Es anagrama"
  else:
    return "No es anagrama"
```

```python
#Ejercicio 6
def buscar_nombre():
  lista_nombres = input("Introduce la lista de nombres separador por comas: ")
  lista_nombres=list(lista_nombres.split(","))
  nombre_buscar = input("Introduce el nombre que quieres buscar: ")
  if nombre_buscar in lista_nombres:
    return "Fue encontrado el nombre"
  else:
    raise ValueError("El nombre no se encuentra en la lista")
```

```python
#Ejercicio 7
def fibonacci(n):
  tot=[]
  for e in range(n):
    if e==0:
      tot.append(0)
    elif e==1:
      tot.append(1)
    else:
      calculo= tot[-1]+tot[-2]
      tot.append(calculo)
  return tot
```

```python
#Ejercicio 8
def encontrar_puesto_empleado(nombre_completo,lista_empleados):
  nombre_completo = list(nombre_completo.split(" "))
  nombre = nombre_completo[0]
  apellido = nombre_completo[1]
  for e in lista_empleados:
    if e['nombre'] == nombre and e['apellido'] == apellido:
      return e['puesto']
  return "La persona no trabaja aqui"
```

```python
#Ejercicio 14
class Arbol:
  def __init__(self):
    self.tronco=1
    self.ramas=[]
  def crecer_tronco(self):
    self.tronco=self.tronco+1
  def nueva_rama(self):
    self.ramas.append(1)
  def crecer_ramas(self):
    lista_nueva=[]
    for e in self.ramas:
      lista_nueva.append(e+1)
    self.ramas=lista_nueva
  def quitar_rama(self,posicion):
    self.ramas.pop(posicion)
  def info_arbol(self):
    print(f"La logitud del tronco es de: {self.tronco}")
    print(f"El númeo de ramas es de: {len(self.ramas)}")
    for e in range(len(self.ramas)):
      print(f"En la rama {e} su logitud es de {self.ramas[e]}")
      
```

```python
#Ejercicio 15
class UsuarioBanco:
  def __init__(self,nombre,saldo,cuenta_corriente):
    self.nombre=nombre
    self.saldo=saldo
    self.cuenta_corriente=cuenta_corriente
  def retirar_dinero(self,cantidad):
    if self.saldo < cantidad:
      raise ValueError("No se puede retirar esa cantidad, saldo insuficiente")
    self.saldo=self.saldo-cantidad
  def transferir_dinero(self,otro_usuario, cantidad):
    if not otro_usuario.cuenta_corriente or not self.cuenta_corriente:
      raise ValueError("Ambos usuarios deben de tener cuenta")
    if otro_usuario.saldo < cantidad:
      raise ValueError("El usuario de origen no dispone de suficiente saldo")
    otro_usuario.saldo=otro_usuario-cantidad
    self.saldo=self.saldo+cantidad
  def agregar_dinero(self,cantidad):
    self.saldo=self.saldo+cantidad
```

```python
#Ejercicio 16
class Procesar_texto:
  def __init__(self,texto):
    self.texto=texto
  def contar_palabras(self):
    diccionario={}
    texto_nuevo = self.texto.replace(".","")
    lista_palabras = texto_nuevo.split(" ")
    for e in lista_palabras:
      if e not in diccionario:
        contador=0
        for i in lista_palabras:
          if e==i:
            contador+=1
        diccionario[e]=contador
    return diccionario
  def reemplazar_palabras(self, palabra_original,palabra_nueva):
    self.texto = self.texto.replace(palabra_original,palabra_nueva)
    return self.texto
  def eliminar_palabra(self,palabra_eliminar):
    self.texto =self.texto.replace(palabra_eliminar,"")
    return self.texto
  def procesar_texto(self,accion,*args):
    if accion == "contar":
      return self.contar_palabras()
    elif accion == "reemplazar":
      return self.reemplazar_palabras(*args)
    elif accion == "eliminar":
      return self.eliminar_palabra(*args)
    else:
      raise ValueError("Acción no válida")
 
```
